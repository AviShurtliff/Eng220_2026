```python
n = float(input("Please input a number: "))
m = float(input("Please input a second number: "))
op = input("Choose an operation (+, -, *, /): ")
def add(n,m):
  return n+m
def sub(n,m):
  return n-m
def mul(n,m):
  return n*m
def div(n,m):
  if m == 0:
    return "Invalid: cannot divide by zero"
  return n/m
def operations(op, n, m):
  switch = {
   '+': add,
   '-': sub,
   '*': mul,
   '/': div,
  }
  func = switch.get(op, 'Invalid Input')
  if func is None:
      return "Invalid operation"
  return func(n, m)

result = operations(op, n, m)
print("Result:", result)
```
